## Changements par rapport aux versions sources

Les listes de crédits et d'échéances de retard sont analysées comme des listes, et non converties directement en nombres. Les NA des postes financiers contribuent pour zéro (hypothèse du notebook), ce qui préserve les autres montants connus. Le jour et l'heure du RDV, la civilité, le reste à vivre et la présence d'un découvert sont ajoutés.

L'horizon retraite renseigné est conservé ; les valeurs absentes sont complétées avec l'hypothèse de 64 ans, puis la médiane du train. Les horizons négatifs sont ramenés à zéro.

Les identifiants, `statut_final`, les textes libres et dates brutes sont exclus des prédicteurs. La disponibilité réelle de chaque variable à la prise du rendez-vous reste à confirmer. Les découpages aléatoires répétés des sources ne sont pas conservés : leurs scores ne sont donc pas directement comparables aux résultats présents.

Les courbes et scores sont calculés à partir des prédictions, sans valeurs saisies à la main. Les paramètres de l'arbre sont limités pour faciliter sa lecture ; ils ne sont pas optimisés sur le test.
